## 2.5. Git et GitHub

Ce sont deux choses différentes, et la confusion est constante.

- **Git** est un logiciel qui tourne sur votre machine. Il enregistre l'histoire
  de vos fichiers : qui a changé quoi, quand, et pourquoi.
- **GitHub** est un site web qui héberge des dépôts git, pour les partager et
  travailler à plusieurs.

Git fonctionne parfaitement sans GitHub. GitHub sans git n'aurait aucun sens.

### Le problème que git résout

Vous connaissez déjà ce dossier :

```
analyse.ipynb
analyse_v2.ipynb
analyse_v2_final.ipynb
analyse_v2_final_VRAIMENT.ipynb
analyse_v2_final_corrections_marie.ipynb
```

Personne ne sait plus lequel fait foi, ni ce qui distingue les deux derniers, ni
comment récupérer un bout de code supprimé mardi. Git remplace tout cela par
**un seul fichier** et une histoire consultable.

Ce que git vous donne concrètement :

- revenir à n'importe quel état antérieur ;
- savoir quelle ligne a changé, quand, et pour quelle raison ;
- travailler à plusieurs sans s'écraser mutuellement ;
- montrer votre démarche, pas seulement votre résultat final.

Ce dernier point est évalué dans ce cours. Un devoir livré en un seul commit
« devoir terminé » la veille de l'échéance ne montre aucune démarche. Une suite
de commits qui raconte l'exploration, les impasses et les corrections, oui.

### Le vocabulaire minimal

| Terme | Ce que c'est |
|---|---|
| **repository** (dépôt) | Le projet suivi par git, avec toute son histoire |
| **commit** | Une photo de vos fichiers à un instant, avec un message qui l'explique |
| **push** | Envoyer vos commits vers GitHub |
| **pull** | Récupérer les commits des autres depuis GitHub |
| **clone** | Copier un dépôt GitHub sur votre machine, la première fois |
| **branch** | Une ligne de travail parallèle, pour ne pas casser la principale |

Le cycle de travail est toujours le même :

```
je modifie  ->  git add  ->  git commit -m "message"  ->  git push
```

`add` choisit ce qui entre dans la photo, `commit` prend la photo, `push`
l'envoie sur GitHub.

![Le cycle add, commit, push](https://raw.githubusercontent.com/cvandekerckh/mlsmm2151/main/images/ch02-add-commit-push.png)

*Les quatre endroits où vit votre travail, et les trois commandes qui le font avancer.*


### Écrire un message de commit

Le message explique **pourquoi**, pas **quoi**. Le « quoi » est déjà dans le
diff, git le connaît mieux que vous.

| À éviter | Pourquoi | Mieux |
|---|---|---|
| `update` | Ne dit rien | `Corrige le parsing du prix, qui contenait un $` |
| `fix` | Fix de quoi ? | `Retire les annonces sans prix avant l'entraînement` |
| `travail du 14/11` | La date est déjà dans git | `Ajoute l'analyse des manquants par commune` |
| `ca marche enfin!!!` | Illisible dans six mois | `Corrige la fuite de données due à estimated_revenue` |

La règle pratique : votre message doit compléter la phrase *« ce commit va... »*.
Verbe à l'infinitif ou au présent, une ligne, en français ou en anglais mais pas
les deux.

## 2.6. Travailler avec GitHub depuis Colab

Vous avez deux chemins. Le premier suffit pour tout ce cours.

### Chemin 1 : le bouton de Colab

Colab sait écrire directement dans un dépôt GitHub.

3. Choisissez le dépôt, la branche `main`, le nom du fichier.
4. **Écrivez un vrai message de commit**, pas celui proposé par défaut.

C'est un vrai commit, avec un vrai message. Rien de moins bien qu'en ligne de
commande.

**Limite** : un fichier à la fois. Pour modifier plusieurs fichiers d'un coup,
il faut le chemin 2.

### Chemin 2 : git en ligne de commande

Plus général, et indispensable dès que vous touchez plusieurs fichiers, par
exemple pour rendre à la fois votre notebook et votre `REFLEXION.md`.

Dans une cellule Colab, les commandes système se préfixent d'un `!` :

```python
!git clone https://github.com/VOTRE-COMPTE/VOTRE-DEPOT.git
%cd VOTRE-DEPOT

!git config user.email "prenom.nom@student.uclouvain.be"
!git config user.name "Prenom Nom"

# ... vous modifiez des fichiers ...

!git add .
!git commit -m "Ajoute l'analyse des valeurs manquantes"
!git push
```

Au `push`, GitHub demandera une authentification. **Votre mot de passe GitHub ne
fonctionne pas** : il faut un *personal access token*. Sur GitHub :
`Settings` > `Developer settings` > `Personal access tokens` >
`Tokens (classic)` > `Generate new token`, avec la portée `repo`.

Un token est un mot de passe. Ne le collez jamais dans une cellule que vous
poussez ensuite sur GitHub : vous publieriez la clé de votre propre compte.
Utilisez `getpass`, qui masque la saisie.

```python
from getpass import getpass
token = getpass("Token GitHub : ")
```

### Ce que git ne doit pas voir

Rappel du notebook précédent : le `.gitignore`. Vérifiez toujours **avant** de
commiter ce que vous êtes sur le point d'envoyer.

```python

```

Si vous voyez un `.csv` dans `git status`, arrêtez-vous : votre `.gitignore` est

## 2.8. Collaborer : branche, pull request, relecture

Jusqu'ici, vous travaillez seul dans votre dépôt de devoir, et vous poussez
directement sur `main`. Dès qu'on est plusieurs, ce n'est plus tenable :
personne ne veut qu'un commit d'un collègue, non relu, casse la version qui
fait foi.

La réponse, dans toutes les équipes qui utilisent git, tient en trois mots.

- **Branche.** Vous travaillez sur une copie parallèle de `main`, qui ne gêne
  personne.
- **Pull request.** Quand c'est prêt, vous demandez que votre branche soit
  fusionnée dans `main`. GitHub affiche le diff, et ouvre une discussion.
- **Relecture.** Quelqu'un d'autre lit le diff, commente, puis approuve. Ce
  n'est qu'après qu'on fusionne.
