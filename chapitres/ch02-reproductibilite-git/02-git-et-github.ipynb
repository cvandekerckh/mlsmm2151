## 2.5. Git et GitHub

Ce sont deux choses différentes, et la confusion est constante.

- **Git** est un logiciel qui tourne sur votre machine. Il enregistre l'histoire
  de vos fichiers : qui a changé quoi, quand, et pourquoi.
- **GitHub** est un site web qui héberge des dépôts git, pour les partager et
  travailler à plusieurs.

Git fonctionne parfaitement sans GitHub. GitHub sans git n'aurait aucun sens.

### Le problème que git résout

Vous connaissez déjà ce dossier :

```
analyse.ipynb
analyse_v2.ipynb
analyse_v2_final.ipynb
analyse_v2_final_VRAIMENT.ipynb
analyse_v2_final_corrections_marie.ipynb
```

Personne ne sait plus lequel fait foi, ni ce qui distingue les deux derniers, ni
comment récupérer un bout de code supprimé mardi. Git remplace tout cela par
**un seul fichier** et une histoire consultable.

Ce que git vous donne concrètement :

- revenir à n'importe quel état antérieur ;
- savoir quelle ligne a changé, quand, et pour quelle raison ;
- travailler à plusieurs sans s'écraser mutuellement ;
- montrer votre démarche, pas seulement votre résultat final.

Ce dernier point est évalué dans ce cours. Un devoir livré en un seul commit
« devoir terminé » la veille de l'échéance ne montre aucune démarche. Une suite
de commits qui raconte l'exploration, les impasses et les corrections, oui.

### Le vocabulaire minimal

| Terme | Ce que c'est |
|---|---|
| **repository** (dépôt) | Le projet suivi par git, avec toute son histoire |
| **clone** | Copier un dépôt GitHub sur votre machine, la première fois |
| **commit** | Une photo de vos fichiers à un instant, avec un message qui l'explique |
| **push** | Envoyer vos commits vers GitHub |
| **pull** | Récupérer depuis GitHub ce qui y a changé |

Le cycle de travail est toujours le même :

```
je modifie des fichiers  ->  commit, avec un message  ->  push
```

`commit` prend la photo, `push` l'envoie sur GitHub. Dans ce cours, vous ne
taperez aucune commande : **GitHub Desktop** fait ces deux gestes avec deux
boutons, et vous montre avant chaque photo exactement ce qui va y entrer.

![Le cycle de travail du devoir](https://raw.githubusercontent.com/cvandekerckh/mlsmm2151/main/images/ch02-cycle-devoir.png)

*Trois endroits et quatre gestes : c'est tout ce qu'il faut pour rendre un devoir.*

### Écrire un message de commit

Le message explique **pourquoi**, pas **quoi**. Le « quoi » est déjà dans le
diff, git le connaît mieux que vous.

| À éviter | Pourquoi | Mieux |
|---|---|---|
| `update` | Ne dit rien | `Corrige le parsing du prix, qui contenait un $` |
| `fix` | Fix de quoi ? | `Retire les annonces sans prix avant l'entraînement` |
| `travail du 14/11` | La date est déjà dans git | `Ajoute l'analyse des manquants par commune` |
| `ca marche enfin!!!` | Illisible dans six mois | `Corrige la fuite de données due à estimated_revenue` |

La règle pratique : votre message doit compléter la phrase *« ce commit va... »*.
Verbe à l'infinitif ou au présent, une ligne, en français ou en anglais mais pas
les deux.

## 2.6. Le cycle de travail du devoir : Colab exécute, GitHub Desktop enregistre

Votre notebook de devoir a besoin de deux choses qui ne sont pas au même
endroit : une machine pour l'exécuter, et un dépôt git pour en garder
l'histoire. Colab fournit la première. Mais une session Colab ne conserve
rien, vous l'avez vu au notebook précédent : ce qui n'est pas enregistré
ailleurs est perdu quand elle se ferme.

Le dépôt, lui, vit sur **votre ordinateur**, dans un dossier que GitHub Desktop
a copié depuis GitHub (un *clone*). Une séance de travail enchaîne donc quatre
gestes, toujours les mêmes :

1. **Importer** `notebook.ipynb` dans Colab, depuis ce dossier.
2. **Travailler** dans Colab : écrire, exécuter, corriger.
3. **Télécharger** le notebook (`Fichier` > `Télécharger` > `.ipynb`) dans le
   dossier, à la place de l'ancien.
4. Dans GitHub Desktop : un **commit** avec un message, puis **push**.

Pourquoi ce détour, plutôt qu'un bouton « enregistrer » ? Parce que vous voyez
chaque fichier avant qu'il n'entre dans le dépôt, que `REFLEXION.md` suit
exactement le même chemin que le notebook, et que c'est le geste que vous
referez au deuxième quadrimestre, avec VS Code à la place de Colab. Le
[tutoriel du devoir 1](https://github.com/cvandekerckh/mlsmm2151/blob/main/devoirs/devoir-1-preparation/tutoriel.md)
vous le fait faire pas à pas dans quelques minutes.

### Ce que git ne doit pas voir

Rappel du notebook précédent : le `.gitignore` dit à git ce qu'il doit
ignorer, et celui de votre dépôt de devoir exclut déjà les fichiers de
données. Mais vérifiez toujours **avant** de commiter : la liste `Changes` de
GitHub Desktop montre chaque fichier qui entrera dans la photo. Si un `.csv`
y apparaît, arrêtez-vous : ne le cochez pas, et supprimez-le du dossier. Il est
beaucoup plus facile de ne pas commiter un fichier que de le retirer de
l'historique après coup.

### La limite : GitHub n'est pas fait pour les notebooks

Git a été conçu pour du texte : des lignes qu'on ajoute, qu'on retire, qu'on
compare. Un `.py` est du texte. Un `.ipynb` est un fichier JSON qui contient
vos cellules, mais aussi leurs **sorties** : tableaux, figures encodées en
texte, compteurs d'exécution. Deux conséquences, que vous verrez dès votre
premier commit.

- **Le diff est illisible.** Changer une phrase, ou simplement relancer une
  cellule, produit des centaines de lignes modifiées. GitHub affiche le
  notebook rendu, mais pas une comparaison lisible de deux versions.
- **Le dépôt grossit.** Une figure dans une sortie pèse quelques centaines de
  kilo-octets, et chaque version reste dans l'historique.

Pour un devoir, c'est acceptable : un notebook, un auteur, un lecteur, qui
le relance de toute façon depuis zéro. Mais retenez la raison pour laquelle, en
entreprise, le code stabilisé quitte le notebook pour un `.py` : c'est lui qui
se relit, se compare et se partage.

## 2.7. Rendre un devoir sur GitHub

Les trois devoirs du cours se rendent sur GitHub. Le mécanisme est toujours le
même.

1. À chaque devoir, je crée pour vous **un dépôt privé à votre nom**,
   `mlsmm2151-uclouvain/devoir-N-votre-pseudo`, pré-rempli avec le squelette du
   devoir. C'est à cela que sert le pseudo GitHub donné à l'inscription.
2. Vous recevez une **invitation** à ce dépôt, par e-mail et sur github.com.
   Acceptez-la. Seuls vous et l'enseignant y avez accès.
3. Vous travaillez dedans et vous poussez vos commits au fur et à mesure.
4. **Il n'y a pas de bouton « rendre ».** Ce qui est dans votre dépôt à l'heure
   de l'échéance constitue votre rendu. C'est l'heure à laquelle GitHub reçoit
   votre push qui compte, pas la date inscrite dans le commit.

Ce que contient chaque dépôt de devoir :

```
README.md          L'énoncé et la checklist de remise
notebook.ipynb     Le squelette à compléter
REFLEXION.md       Obligatoire : votre démarche, vos choix, vos limites
sujet.py           Tire votre variante individuelle à partir de votre matricule
requirements.txt
.gitignore         Exclut les données. Ne le modifiez pas
```

Trois points qui coûtent des points chaque année :

- **`REFLEXION.md` est obligatoire.** Un devoir rendu sans ce fichier n'est pas
  corrigé.
- **Au moins cinq commits** répartis dans le temps. L'historique fait partie de
  ce qui est évalué.
- **Aucun fichier de données** dans le dépôt. Votre notebook doit télécharger ce
  dont il a besoin.

## 2.8. Pour aller plus loin : branches, pull requests, et le dossier `students/`

Dans votre dépôt de devoir, vous êtes seul : vous poussez directement sur
`main`, et c'est très bien ainsi. Dès qu'on est plusieurs sur un même dépôt,
les équipes ajoutent deux outils que vous n'avez pas besoin de maîtriser pour
ce cours, mais dont vous entendrez parler : la **branche**, une ligne de travail
parallèle qui ne gêne personne, et la **pull request**, par laquelle on demande
que cette branche soit relue puis fusionnée dans `main`.

C'est ainsi que fonctionne le dossier
[`students/`](https://github.com/cvandekerckh/mlsmm2151/tree/main/students)
du dépôt du cours, où sont publiées les notes collectives de chaque séance :
tout passe par une pull request, approuvée par un autre étudiant, et GitHub
guide chaque étape depuis son site, sans rien installer. Le mode d'emploi est
dans [`CONTRIBUTING.md`](https://github.com/cvandekerckh/mlsmm2151/blob/main/CONTRIBUTING.md) :
le responsable du pad de la séance s'en sert pour publier les notes.

> **La règle du dépôt du cours : en dehors de `students/`, vous ne touchez à
> rien.** Le matériel (chapitres, devoirs, README) n'est modifié que par
> l'enseignant, et une pull request qui touche un fichier hors de `students/`
> est refusée automatiquement.

## À retenir

- **Git** enregistre l'histoire de vos fichiers, **GitHub** l'héberge et la
  partage.
- Le cycle : modifier, commit avec un message qui dit « pourquoi », push.
- Un message de commit complète la phrase *« ce commit va... »*.
- Le devoir se travaille en quatre gestes : importer dans Colab, travailler,
  télécharger dans le dossier du dépôt, commit et push dans GitHub Desktop.
  Avant de fermer Colab : télécharger, toujours.
- Il n'y a pas de bouton « rendre » : ce qui est dans le dépôt à l'heure de
  l'échéance est votre rendu. Cinq commits au moins, `REFLEXION.md`
  obligatoire, aucune donnée.
- Regardez la liste `Changes` avant chaque commit. Un `.csv` qui y apparaît
  est un signal d'alarme.
- GitHub compare mal les notebooks : acceptable pour un devoir, pas pour du
  code qu'on partage, qui va dans un `.py`.
- Dans le dépôt du cours, vous n'écrivez que dans `students/`.

**La suite** : les outils sont en place. Le chapitre 3 attaque les données
elles-mêmes : explorer, diagnostiquer, nettoyer, traiter le temps. C'est la
matière du devoir 1, que vous démarrez en fin de séance, dans le dépôt que
